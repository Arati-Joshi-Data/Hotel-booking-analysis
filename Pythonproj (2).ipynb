```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv('hotel_bookings.csv')

print("Dataset shape:", df.shape)
df.head()
```

```python
df['children'] = df['children'].fillna(0)
df.isnull().sum()
```

```python
df['total_nights'] = (
    df['stays_in_weekend_nights'] +
    df['stays_in_week_nights']
)

df['total_guests'] = (
    df['adults'] +
    df['children'] +
    df['babies']
)

df['estimated_revenue'] = (
    df['adr'] * df['total_nights']
)

df[
    ['total_nights',
     'total_guests',
     'estimated_revenue']
].head()
```

```python
features = [
    'lead_time',
    'total_nights',
    'total_guests',
    'is_repeated_guest',
    'previous_cancellations',
    'previous_bookings_not_canceled',
    'booking_changes',
    'adr',
    'required_car_parking_spaces',
    'total_of_special_requests'
]

X = df[features].copy()

X.head()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Data scaled successfully.")
```

```python
inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    range(2, 11),
    inertia,
    marker='o'
)

plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method - Finding Optimal Customer Segments')

plt.show()
```

```python
k = 4

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

df['Segment'] = kmeans.fit_predict(X_scaled)

df['Segment'].value_counts().sort_index()
```

```python
segment_analysis = df.groupby('Segment').agg(
    Customers=('Segment', 'count'),
    Avg_Lead_Time=('lead_time', 'mean'),
    Avg_Total_Nights=('total_nights', 'mean'),
    Avg_Total_Guests=('total_guests', 'mean'),
    Repeat_Guest_Rate=('is_repeated_guest', 'mean'),
    Avg_Cancellations=('previous_cancellations', 'mean'),
    Avg_Previous_Bookings=('previous_bookings_not_canceled', 'mean'),
    Avg_Booking_Changes=('booking_changes', 'mean'),
    Avg_ADR=('adr', 'mean'),
    Avg_Special_Requests=('total_of_special_requests', 'mean'),
    Avg_Revenue=('estimated_revenue', 'mean')
)

segment_analysis
```
